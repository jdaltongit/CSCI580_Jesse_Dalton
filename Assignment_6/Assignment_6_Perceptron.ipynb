```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data-1.csv', header=None, names =['x1','x2','label'])

class1 = df[df['label'] == 1]
class0 = df[df['label'] == 0]

plt.scatter(class1['x1'], class1['x2'], color = 'blue', label = '1')
plt.scatter(class0['x1'], class0['x2'], color = 'red', label = '0')
```

```python
# Part 1 - Heuristic Perceptron (lr = 0.01)

# Step Function
def step(x):
  return 1 if x >= 0 else 0

def plot_line(w1, w2, b, color, linestyle = '-'):
  x1_vals = np.linspace(0, 1, 100)
  x2_vals = (-w1 * x1_vals - b) / w2
  plt.plot(x1_vals, x2_vals, color = color, linestyle = linestyle)

X = df[['x1', 'x2']].values
y = df['label'].values

learning_rate = 0.01
iterations = 65

# Step 1: Start with random weights
w1, w2, b = np.random.randn(), np.random.randn(), np.random.randn()

plt.figure(figsize=(6,6))
plt.scatter(class1['x1'], class1['x2'], color = 'blue', label = '1')
plt.scatter(class0['x1'], class0['x2'], color = 'red', label = '0')

# Plot initial line
plot_line(w1, w2, b, color = 'red')

# Step 2: For each of all points (data) with their corresponding labels (answers):
for it in range(iterations):
    for i in range(len(X)):
        x1_i, x2_i = X[i]
        y_i = y[i]

        # Step 2.1: Classify according to the perceptron
        y_hat = step(w1 * x1_i + w2 * x2_i + b)

        # Step 2.2: For a misclassified point
        if y_hat != y_i:
            if y_hat == 0:
                # Step 2.2.1
                b  = b  + learning_rate
                w1 = w1 + learning_rate * x1_i
                w2 = w2 + learning_rate * x2_i
            else:
                # Step 2.2.2
                b  = b  - learning_rate
                w1 = w1 - learning_rate * x1_i
                w2 = w2 - learning_rate * x2_i

    # Plot green iteration lines
    if it < iterations - 1:
            plot_line(w1, w2, b, color='green', linestyle='--')

# Plot final line in black
plot_line(w1, w2, b, color='black')

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.title(f'Solution boundary (lr={learning_rate})')
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()

print(f"Final weights: w1={w1:.4f}, w2={w2:.4f}, b={b:.4f}")
```

```python
# Part 2 - Gradient Descent Perceptron (lr = 0.01, epochs = 100)

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

learning_rate = 0.01
epochs = 100

# Step 1: Start with random weights
w1, w2, b = np.random.randn(), np.random.randn(), np.random.randn()

plt.figure(figsize=(6,6))
plt.scatter(class1['x1'], class1['x2'], color='blue', label='1')
plt.scatter(class0['x1'], class0['x2'], color='red', label='0')

# Plot initial red line
plot_line(w1, w2, b, color='red')

error_history = []

# Step 2: For each of all points with their corresponding labels
for epoch in range(epochs):
    for i in range(len(X)):
        x1_i, x2_i = X[i]
        y_i = y[i]

        # Step 2.1
        y_hat = sigmoid(w1 * x1_i + w2 * x2_i + b)

        # Step 2.2
        error = y_i - y_hat

        # Step 2.3
        b  = b  + learning_rate * error

        # Step 2.4
        w1 = w1 + learning_rate * error * x1_i
        w2 = w2 + learning_rate * error * x2_i

    # Plot green iteration lines
    if epoch < epochs - 1:
        plot_line(w1, w2, b, color='green', linestyle='--')

    # Compute log loss every 10 epochs
    if epoch % 10 == 0:
        y_hat_all = sigmoid(w1 * X[:, 0] + w2 * X[:, 1] + b)
        log_loss = -np.mean(y * np.log(y_hat_all) + (1 - y) * np.log(1 - y_hat_all))
        error_history.append((epoch, log_loss))

# Plot final line in black
plot_line(w1, w2, b, color='black')

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.title(f'Solution boundary (lr={learning_rate})')
plt.xlabel('x1')
plt.ylabel('x2')
plt.show()

# Plot log loss error graph
epochs_list = [e[0] for e in error_history]
loss_list   = [e[1] for e in error_history]

plt.figure(figsize=(7, 4))
plt.plot(epochs_list, loss_list)
plt.title(f'Error Plot (lr={learning_rate})')
plt.xlabel('Number of epochs')
plt.ylabel('Error')
plt.show()

print(f"Final weights: w1={w1:.4f}, w2={w2:.4f}, b={b:.4f}")
```
